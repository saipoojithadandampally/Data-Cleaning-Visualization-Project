```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
```

```python

print("First 5 rows:")
display(df.head())

print("Dataset dimensions:")
print(df.shape)

print("Column names:")
print(df.columns.tolist())

print("Dataset information:")
df.info()
```

```python

display(df.describe(include="all"))
```

```python

# Count missing values in each column

missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)

# Calculate missing value percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100

print("\nMissing Value Percentage:")
print(missing_percentage.round(2))
```

```python

plt.figure(figsize=(12, 6))

sns.heatmap(df.isnull(), cbar=False, cmap="viridis")

plt.title("Missing Values Heatmap")
plt.show()
```

```python

print("Total duplicate rows:", df.duplicated().sum())

# Remove duplicate records
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python

# Fill missing Age values with median
df['age'] = df['age'].fillna(df['age'].median())

# Fill missing Embarked values with mode
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])

# Drop Cabin column
df = df.drop(columns=['deck'])

print("Missing values handled successfully!")

print(df.isnull().sum())
```

```python

print("Remaining missing values:")
print(df.isnull().sum())

print("\nCleaned Dataset:")
display(df.head())
```

```python

plt.figure(figsize=(12, 5))

sns.boxplot(data=df[['age', 'fare']])

plt.title("Outlier Detection Using Boxplot")
plt.xlabel("Features")

plt.show()
```

```python

columns = ['age', 'fare']

for col in columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]

    print(f"\nColumn: {col}")
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Number of Outliers:", len(outliers))
```

```python

df_cleaned = df.copy()

for col in ['age', 'fare']:
    Q1 = df_cleaned[col].quantile(0.25)
    Q3 = df_cleaned[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    df_cleaned[col] = df_cleaned[col].clip(
        lower=lower_limit,
        upper=upper_limit
    )

print("Outlier treatment completed!")
```

```python

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

sns.boxplot(x=df['age'], ax=axes[0, 0])
axes[0, 0].set_title("Age Before Treatment")

sns.boxplot(x=df_cleaned['age'], ax=axes[0, 1])
axes[0, 1].set_title("Age After Treatment")

sns.boxplot(x=df['fare'], ax=axes[1, 0])
axes[1, 0].set_title("Fare Before Treatment")

sns.boxplot(x=df_cleaned['fare'], ax=axes[1, 1])
axes[1, 1].set_title("Fare After Treatment")

plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(7, 5))

sns.countplot(data=df_cleaned, x='survived', hue='survived')

plt.title("Passenger Survival Analysis")
plt.xlabel("Survival (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()
```

```python

plt.figure(figsize=(7, 5))

sns.countplot(data=df_cleaned, x='sex', hue='survived')

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Passenger Count")

plt.show()
```

```python

plt.figure(figsize=(7, 5))

sns.countplot(data=df_cleaned, x='class', hue='survived')

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Passenger Count")

plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.histplot(data=df_cleaned, x='age', bins=25, kde=True)

plt.title("Passenger Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python

numeric_df = df_cleaned.select_dtypes(include=np.number)

plt.figure(figsize=(10, 6))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Feature Correlation Heatmap")
plt.show()
```

```python

print("========== DATA ANALYSIS REPORT ==========")

print("\n1. Total Passengers:", len(df_cleaned))

print("\n2. Survival Count:")
print(df_cleaned['survived'].value_counts())

print("\n3. Survival Percentage:")
print(
    (df_cleaned['survived'].value_counts(normalize=True) * 100)
    .round(2)
)

print("\n4. Gender Distribution:")
print(df_cleaned['sex'].value_counts())

print("\n5. Average Passenger Age:")
print(round(df_cleaned['age'].mean(), 2))

print("\n6. Average Ticket Fare:")
print(round(df_cleaned['fare'].mean(), 2))

print("\n7. Survival by Gender:")
print(
    df_cleaned.groupby('sex')['survived']
    .mean().mul(100).round(2)
)

print("\n8. Survival by Passenger Class:")
print(
    df_cleaned.groupby('class')['survived']
    .mean().mul(100).round(2)
)
```

```python

df_cleaned.to_csv(
    'cleaned_titanic_dataset.csv',
    index=False
)

print("Cleaned dataset saved successfully!")
```
